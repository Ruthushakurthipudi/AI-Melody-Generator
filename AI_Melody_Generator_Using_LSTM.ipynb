```python

import os
import glob
import numpy as np
import matplotlib.pyplot as plt

from music21 import converter, instrument, note, chord, stream
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.utils import to_categorical

print("All libraries imported successfully!")
```

```python

dataset_path = "/content/midi_dataset"

os.makedirs(dataset_path, exist_ok=True)

print("Dataset folder created successfully!")
print("Folder location:", dataset_path)
```

```python

from music21 import corpus

# Find Bach music files
bach_files = corpus.getComposer("bach")

print("Number of music files found:", len(bach_files))

# Display the first 5 files
for file_path in bach_files[:5]:
    print(file_path)
```

```python

from music21 import converter

midi_count = 0

for i, file_path in enumerate(bach_files[:20]):
    try:
        # Read the music file
        score = converter.parse(file_path)

        # Create a MIDI file path
        output_path = os.path.join(
            dataset_path,
            f"music_{i+1}.mid"
        )

        # Save the music as MIDI
        score.write("midi", fp=output_path)

        midi_count += 1

    except Exception as error:
        print(f"Skipped file {i+1}: {error}")

print("MIDI files created:", midi_count)
```

```python

midi_files = glob.glob(
    os.path.join(dataset_path, "*.mid")
)

print("Total MIDI files:", len(midi_files))

for file_path in midi_files[:5]:
    print(file_path)
```

```python

from music21 import converter, instrument, note, chord

notes = []

for file_path in midi_files:
    try:
        # Read the MIDI file
        midi = converter.parse(file_path)

        # Get all notes and chords
        notes_to_parse = None

        parts = instrument.partitionByInstrument(midi)

        if parts:
            notes_to_parse = parts.parts[0].recurse()
        else:
            notes_to_parse = midi.flat.notes

        # Extract notes and chords
        for element in notes_to_parse:
            if isinstance(element, note.Note):
                notes.append(str(element.pitch))

            elif isinstance(element, chord.Chord):
                notes.append(".".join(
                    str(n) for n in element.normalOrder
                ))

    except Exception as error:
        print("Error processing file:", error)

print("Total musical notes and chords:", len(notes))
```

```python

print("First 30 extracted notes and chords:")

print(notes[:30])
```

```python

# Get all unique notes and chords
unique_notes = sorted(set(notes))

# Create mappings
note_to_int = {
    note_name: number
    for number, note_name in enumerate(unique_notes)
}

int_to_note = {
    number: note_name
    for note_name, number in note_to_int.items()
}

print("Total unique notes:", len(unique_notes))
print("First 10 note mappings:")

for item in list(note_to_int.items())[:10]:
    print(item)
```

```python

sequence_length = 40

network_input = []
network_output = []

for i in range(len(notes) - sequence_length):
    sequence_in = notes[i:i + sequence_length]
    sequence_out = notes[i + sequence_length]

    network_input.append([
        note_to_int[item]
        for item in sequence_in
    ])

    network_output.append(note_to_int[sequence_out])

print("Input sequences:", len(network_input))
print("Output values:", len(network_output))
```

```python

n_patterns = len(network_input)

X = np.reshape(
    network_input,
    (n_patterns, sequence_length, 1)
)

# Normalize the input values
X = X / float(len(unique_notes))

# Convert output values to categorical format
y = to_categorical(
    network_output,
    num_classes=len(unique_notes)
)

print("Input shape:", X.shape)
print("Output shape:", y.shape)
```

```python
model = Sequential([
    LSTM(256, input_shape=(X.shape[1], X.shape[2])),
    Dropout(0.3),
    Dense(128, activation="relu"),
    Dense(len(unique_notes), activation="softmax")
])

model.compile(
    loss="categorical_crossentropy",
    optimizer="adam"
)

model.summary()
```

```python
history = model.fit(
    X,
    y,
    epochs=20,
    batch_size=64,
    verbose=1
)
```

```python
import random

start = random.randint(0, len(network_input) - 1)

pattern = network_input[start]
generated_notes = []

for _ in range(100):
    prediction_input = np.reshape(
        pattern, (1, len(pattern), 1)
    )

    prediction_input = prediction_input / float(len(unique_notes))

    prediction = model.predict(
        prediction_input, verbose=0
    )

    index = np.argmax(prediction)
    result = int_to_note[index]

    generated_notes.append(result)

    pattern.append(index)
    pattern = pattern[1:]

print("Generated notes:", len(generated_notes))
print(generated_notes[:10])
```

```python
from music21 import stream, note, chord

output = stream.Stream()

for item in generated_notes:
    if "." in item and item.replace(".", "").isdigit():
        chord_notes = [
            note.Note(int(pitch))
            for pitch in item.split(".")
        ]
        output.append(chord.Chord(chord_notes))
    else:
        output.append(note.Note(item))

output_path = "/content/generated_music.mid"

output.write("midi", fp=output_path)

print("Music saved successfully!")
print(output_path)
```
